## Research Question and Motivation

  For our research project, we are interested in examining the Google Play Store user reviews for the mobile game Clash Royale. Our research question is primarily focused on defining the primary factors behind a positive or negative review on the app store. The motivation for this stems primarily from an interest in mobile gaming as an industry, and seeing if information obtained from player reviews can provide a useful basis for proposing changes and updates to a game to better service players and popularize a game.

  Considering these goals, a successful project should be able to find a meaningful, substantive relationship between commentary and review score that could be interpreted to plan and implement specific changes to a game such as might be proposed to game designers or company executives.

## Data

###Dataset Acquisition

Our dataset consists of roughly 2.3 million Clash Royale user reviews scraped using the open source "google-play-scraper" (https://pypi.org/project/google-play-scraper/).

The dataset can be viewed and downloaded here:
https://drive.google.com/file/d/1nlbmyhJU6A5Uu5pmJoR7VcBj5_-XobwB/view?usp=sharing

###Data Description
The dataset contains a total of 2,307,096 unique reviews, with 6 different columns:
  - review_id: The unique identification number of each review by the Google Play Store
  - content: The review's "description," or specific commentary by the user, which is our target for analysis
  - score: The star rating (on a 5-star scale) of the app
  - at: This is the date and time that the review was created
  - likes: The number of likes that the review has received
  - appVersion: The version of the app that the review was uploaded for

### Ethical Concerns
There are no significant ethical concerns with the dataset as all the data acquired is publicly avaliable through the Google Play Store. Reviews are intentionally published by users and expected to be publicly viewed.

## Problem Setup

Our primary focus for this project is a sentiment analysis, looking at whether a review for the game is considered "positive" or "negative," rather than the specific number of stars that the review has. Because of this, we filtered the "score" data such that a review with 3, 4, or 5 stars is considered to be a positive review, whereas a review with a score of 2 or below is considered negative. We make use of logistic regression models to determine what words in the content of a review are most significant in determining a positive and negative review.

Our target variable in this project is the content of the reviews, as we are interested in seeing the specific commentary that is associated with positive and negative sentiments. Factoring this decision into our data, we cut out reviews that are blank and do not have anything in the "content" section, which cuts down our dataset to 2,269,390 reviews.

In terms of our assumptions read into the data, the most impactful decision we made is defining 3-star reviews as carrying a "positive" sentiment, though we recognize that there is room for interpretation and debate here. Considering that the vast majority of reviews are either 1 or 5 stars (2 and 3 stars make up a very small percentage of the overall data), however, it is unlikely to have a significant impact overall on the quality and validity of our results and analysis.



## Methods

###Baseline Approach
Because our focus for this project is a text-as-data sentiment analysis, and our data was fairly simply categorized into two opposing categories, we chose a fairly simplistic model for our initial investigation: a "bag of words" style single-word analysis

For the bag of words model, we were primarily interested in whether or not any specific comments stood out as potential explanations for the review sentiment, as it would make sense that certain descriptive words might appear more frequently in positive or negative reviews. After removing stopwords (specifically the 318-word list in the sklearn.feature_extraction.text's stopwords package), we extracted the top 20 most common words from positive reviews and the top 20 most common words from negative reviews

###Final Models: Temporal Logistic Regression and Topic Model
####Logistic Regression
Ultimately, for our conclusive model, we designed a temporal logistic regression, including four time periods which we qualitatively determined based on the major update history of the game. These "eras" of the game's life cycle are as follows:
  - Prior to October 1st, 2021 (Before the release of any particularly major card updates)
  - October 1st, 2021 - June 1st, 2023 (The introduction of "Champion Cards" into the game)
  - June 1st, 2023 - December 1st, 2025 (The introduction of "Card Evolutions" into the game)
  - December 1st, 2025 and onward (The introduction of "Heros" into the game)

We then created four data samples (one for each era) with equally-sampled positive and negative reviews, and created a logistic regression to determine and extract the 25 strongest positive and negative sentimentally-charged coefficients for given words in the data. ####Topic Model
In addition the the logistic regression model, we included a topic model as a supplimentary approach, hoping to increase the specificity of our analysis while simultaneously confirming findings from our logistic regression model. For these, we created two separate models - one for positive reviews and one for negative reviews (in theory this is not necessary for the topic model approach, but since our goal was to bolster specificity, we modeled it as such to help narrow down topics even further).

In terms of tuning the topic model, we tested a variety of min_df and max_df values to see how it would impact the output's interpretability and value, settling on a min_df of 50 documents and a max_df of 90% of documents. Values less than and greater than (for minimum and maximum respectively) our chosen values ended up sacrificing coherence or specificity, hurting our analysis. ##Evaluation and Diagnostics

For our logistic regression analysis, we examined the use of a variety of preprocessing decisions to create the best model, including choices that removed numbers, included and excluded stopwords, and different combinations of the min_df and max_df. None of these merited any impactful change on any of the regressions, only leading to a 4% decrease in overall accuracy when raising the min_df and lowering the max_df. The best diagnostic decision we found was the removal of numbers, stopwords, and a min_df of 0.001 and a max_df of 0.9.

After creating the model as described, we evaluated their validity based on their accuracy and F1 metrics, which for all four era regressions fell around 0.84 to 0.86 for both for overall accuracy and overall f1 score, tending to perform best for negative review precision and positive review recall:

## Interpretation and Substantive Takeaways

Looking at the data wholistically, there are a few fairly strong conclusions that we can draw relating to specific features of Clash Royale. Primarily, from the logistic regression, we can see a significant increase over time in the mention of monetization of the game within the negative reviews - pre-2021 it's not mentioned at all in the coefficients, but by post-2025 there are 9 coefficients that can be reasonably inferred (if not explicitly stated) to be related to the money one might spend on the game. Aside from this, we can see other meaningful complaints from the topic model, the most notable being balancing issues in the game, which is mentioned multiple times in both the coefficients and the topic model.

In terms of positive conclusions about the game, there is much less specificity overall within our models, though mentions of the game's "strategic" and "addicting" characteristics are prevalent in both models, and we would conclude that these are the two primary factors behind a player's enjoyment of the game.

## Limitations and Future Work

####Limitations
There are a few limitations in terms of drawing applicable conclusions from our data:
 - Because our data comes entirely from the population of players who actually choose to leave reviews, it may not be representative of player attitudes overall, and it might not fully or accurately reflect the complaints or complements that players have.
 - Overall, the specificity of the model in terms of drawing conclusions about positive reviews is not incredibly strong - many of the displayed words are general words of appraisal that do not offer a lot of valuable substantive information. Aside from the couple mentions of "strategic" and "addictive," there really isn't much useful specific information.
####Potential for Future Work
Were we to continue with this project and our analysis, we would likely do a qualitative, manual assessment of some sample of our reviews to determine whether or not we should exclude many of the generalized words (e.g "awesome," "great," "worst," "terrible") to see if there are more valuable, specific words that pop up instead. Doing this could allow for a more meaningful and substantive analysis of the data and a more useful and applicable interpretation and conclusion, one which might be more appropriate for proposing changes to the game.
